# Solving Equations

```python
from sympy import Symbol, solve 
x = Symbol('x')
expr = x - 5 - 7

v = solve(expr)
print("dict=False")
print(v)
print(v[0])

c = solve(expr, dict=True)
print("\ndict=True")
print(c)
print(c[0][x])
```

# Solving Quadratic Equations

```python
from sympy import solve
from sympy.plotting import plot
x = Symbol('x')
expr = x**2 + 5*x + 4
solve(expr, dict=True) 
# The second argument to the solve() function (dict=True) specifies that we want the result to be returned as a list of Python dictionaries.

plot(expr, (x,-5, 0))   # Noted that you can specify x domain of the plot
```

# Exercise 2: Solving Equation

\begin{align}
s=\frac{1}{2}gt^2
\end{align}
s: Distance or Hight of falling object in meter
\begin{align}
g:=9.8\frac{m}{s^2}
\end{align}
t: time in second

Find out time for falling object at 120 meters from HIGHTOWER. Show graph of falling equation intersec x-axis.

```python
from sympy import solve, Symbol
from sympy.plotting import plot
g = -9.8
t = Symbol("t")
s = 1/2 * g * t**2 + 120
sn = 0
print(o := solve((s), dict=True))
plot(s, (t, 0, 6))
# sn = s
# 0 = s - sn
```

# Solving a System of Linear Equations

\begin{align}
2x+3y=6
\end{align}
\begin{align}
3x+2y=12
\end{align}


```python
from sympy import *
from sympy.plotting import plot
x, y = symbols('x,y')
expr1 = 2*x + 3*y - 6
expr2 = 3*x + 2*y-12
a = solve((expr1, expr2), dict=True)
print(a)
print(solve((expr1, expr2), dict=True))
pexp1= (-2*x+6)/3
pexp2 = (-3*x +12)/2
plot(pexp1, pexp2)
```
