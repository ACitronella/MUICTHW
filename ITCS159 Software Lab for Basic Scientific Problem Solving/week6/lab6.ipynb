```python
import scipy
from scipy.cluster.vq import kmeans, vq, whiten
```

```python
student_scores = [[80] , [78] , [55] , [60] , [62] ,
                [59] , [48] , [49] , [51] , [54] ,
                [72] , [95] , [78] , [54] , [50] ,
                [61] , [57] , [56] , [65] , [53]];
```

```python
student_scores = whiten(student_scores) # normalize 
centroids ,_ = kmeans(student_scores, 3) # mean calculate algo
```

```python
student_scores
```

```python
centroids
```

```python
result ,_ = vq(student_scores, centroids)
```

```python
# seem result is telling no. group of each in student scores
result
```

```python
result.reshape(4, 5) # group 1 is bigger than group 0
```

```python
student_scores = [[80] , [78] , [55] , [60] , [62] ,
                [59] , [48] , [49] , [51] , [54] ,
                [72] , [95] , [78] , [54] , [50] ,
                [61] , [57] , [56] , [65] , [53]];
print(stats.tmean(student_scores)) # mean
print(stats.tstd(student_scores)) # sample std
print((62 - stats.tmean(student_scores))/stats.tstd(student_scores)) # z score
```

```python
student_scores = np.array(student_scores)
```

```python
z_all = (student_scores - stats.tmean(student_scores))/stats.tstd(student_scores)
```
