## Series

```python
data = np.array(['a','b','c','d'])
s1 = pd.Series(data)
print(s1)

data = np.array([1,2,3,4])
s2 = pd.Series(data)
print(s2)
```

### What are the results? What are the differences between s1 and s2?

Ans: s1 is series of string, s2 is series of integer

## Dataframe

### Create a DataFrame from Dict of Series

```python
d = {'one' : pd.Series([1, 2, 3], index=['a', 'b', 'c']),
'two' : pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd'
])}
df = pd.DataFrame(d)
print(df)
```

### Notice that we have NaN, why?

NaN exists in this dataframe because one's d is not defined by user.

## Missing Data

## Check for Missing Values

```python
df = pd.DataFrame(np.random.randn(5, 3), index=['a', 'c', 'e','f','h'],columns=['one', 'two', 'three'])
df = df.reindex(['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h'])
print(df['one'].isnull(), end="\n\n")
print(df['two'].notnull())
```

## Filling Missing Data

```python
s1 = reindexed_df.copy()
s2 = reindexed_df.copy()
s1 = s1.fillna(0)
s2 = s2.fillna(method='pad')
print(s1, end="\n\n")
print(s2)
```

### Exercise: What is the differences between s1 and s2.

s1's nan filled with constant value provided. In this case 0.
s2's nan filled with the value before itself
