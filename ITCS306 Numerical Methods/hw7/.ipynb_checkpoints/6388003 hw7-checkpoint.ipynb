## 1. Modify the Gauss's Elimination code provided in the lecture 7 to include part 1 of the row echelon procedure (to normalize the first non-zero coefficient to 1).

```python
import numpy as np
```

```python
def row_echelon(Ab):
    MAb = np.array(Ab, dtype="float64")
    [nr_Ab, nc_Ab] = MAb.shape
    for r in range(nr_Ab): 
        factor = MAb[r, r]
        
        if factor == 0.: # find new factor
            for j in range(r+1, MAb.shape[0]):
                factor = MAb[j, r]
                if factor != 0.: # swap line
                    tmp = MAb[j].copy()
                    MAb[j] = MAb[r]
                    MAb[r] = tmp 
                    break
                    
        MAb[r, :] = (1 / factor) * MAb[r, :]
        for i in range(r+1, nr_Ab):
            row_factor = MAb[i, r]
            tmp_row = row_factor * MAb[r, :].copy()
            MAb[i, :] = MAb[i, :] - tmp_row
    return MAb
```

```python
m = np.array([
    [-1., 3., 2],
    [-1., 21., 4]
], dtype="float64")
```

```python
row_echelon(m)
```

```python
A = [[3,0,2],
     [2,0,-2],
     [0,1,1]]
```

```python
row_echelon(A)
```

```python
m = np.array([
    [1., 2., -1., 1, 6],
    [-1, 1, 2, -1, 3],
    [2, -1, 2, 2, 14],
    [1, 1, -1, 2, 8]
], dtype="float64")
```

```python
row_echelon(m)
```

## 2.Modify the code for Gauss-Jordan Elimination such that you can find the inverse of a matrix A.  A = [[3,0,2],[2,0,-2],[0,1,1]].

```python
def gauss_jordan_elim(a):
    a = np.array(a, dtype="float64")
    for i in range(a.shape[0]):
        factor = a[i, i]
        if factor == 0.: # find new factor
            for j in range(i+1, a.shape[0]):
                factor = a[j, i]
                if factor != 0.: # swap line
                    tmp = a[j].copy()
                    a[j] = a[i]
                    a[i] = tmp 
                    break
        a[i] = a[i] / factor
        for j in range(a.shape[0]):
            if i != j:
                inter_line = a[i] * a[j, i]
                a[j] = a[j] - inter_line
    return a

def gauss_jordan_elim_inv(a):
    a = np.array(a, dtype="float64")
    a = np.insert(a, a.shape[0], np.eye(a.shape[0]), axis=1) # add identity line
    a = gauss_jordan_elim(a) # gauss jordan thingies
    a = a[:, a.shape[0]:] # remove first identity line
    return a
```

```python
gauss_jordan_elim(m)
```

```python
A = [[3,0,2],
     [2,0,-2],
     [0,1,1]]
```

```python
a = gauss_jordan_elim_inv(A)
a
```

```python
np.linalg.inv(A)
```
