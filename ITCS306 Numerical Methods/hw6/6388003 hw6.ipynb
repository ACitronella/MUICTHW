## 1. Find out whether a matrix is a diagonal matrix.  (First check whether the matrix is a square matrix and then check whether the non diagonal elements are zeroes).

```python
import numpy as np
```

```python
def check_diag(m):
    shape = m.shape
    # check rank 2 and square matrix
    if len(shape) == 2 and shape[0] == shape[1]:
        for i in range(shape[0]):
            for j in range(i):
                if m[i, j] != 0 and m[j, i] != 0:
                    return False
        return True
    return False
```

```python
# slower for small m
def check_diag_v2(m):
    shape = m.shape
    return len(shape) == 2 and shape[0] == shape[1] and np.sum(np.abs(m)) == np.sum(np.abs(m.diagonal()))
```

```python
def __check_diag_v3(m, i, j):
    if j == m.shape[1]:
        return __check_diag_v3(m, i+1, 0)
    if i == m.shape[0]-1 and j == m.shape[1]-1:
        return True
    if i == j:
        return __check_diag_v3(m, i, j+1)
    if m[i, j] != 0:
        return False
    return __check_diag_v3(m, i, j+1)
    

def check_diag_v3(m):
    shape = m.shape
    return len(shape) == 2 and shape[0] == shape[1] and __check_diag_v3(m, 0, 0)
```

## 2. Find out whether a matrix is a skew-symmetric matrix.  (First check whether the matrix is a square matrix and then check whether a_ij = - a_ji)

```python
def check_skew_sym(m):
    shape = m.shape
    if len(shape) == 2 and shape[0] == shape[1]:
        for i in range(shape[0]):
            for j in range(i):
                if m[i, j] != -m[j, i]:
                    return False
        return True
    return False
```

```python
def __check_skew_sym_v3(m, i, j):
    if i == m.shape[0]-1 and j == m.shape[1]-1:
        return True
    if i == j:
        return __check_skew_sym_v3(m, i+1, 0)
    if m[i, j] != -m[j, i]:
        return False
    return __check_skew_sym_v3(m, i, j+1)
    

def check_skew_sym_v3(m):
    shape = m.shape
    return len(shape) == 2 and shape[0] == shape[1] and __check_skew_sym_v3(m, 0, 0)
```

```python
m = np.array([
    [1, 2, 3],
    [4, 0, 6],
    [7, 8, 9]
])
m
```

```python
check_skew_sym(m)
```

```python
check_skew_sym_v3(m)
```

## 3. Find an inverse of a matrix by finding its determinant, minor, cofactor and adjugate.

```python
def det(m: np.ndarray):
    global det_count
    if m.shape[0] == 1:
        det_count += 1
        return m[0]
    m = np.copy(m)
    reduce_m = np.delete(m, 0, 0)
    s = 0
    for i in range(m.shape[0]):
        n = np.delete(reduce_m, i, 1)
        s += (-1)**(i) * m[0, i] * det(n)
    return s

def minor(m):
    out = np.zeros(m.shape)
    for i in range(m.shape[0]):
        n = np.delete(m.copy(), i, 0)
        for j in range(m.shape[1]):
            p = np.delete(n.copy(), j, 1)
            out[i, j] = det(p)
    return out

def cofactor(m):
    out = np.zeros(m.shape)
    for i in range(m.shape[0]):
        n = np.delete(m.copy(), i, 0)
        for j in range(m.shape[1]):
            p = np.delete(n.copy(), j, 1)
            out[i, j] = (-1)**(i + j) * det(p)
    return out

def adjoint(m):
    return cofactor(m).T

def inverse_matmul(m):
    if len(m.shape) != 2 or m.shape[0] != m.shape[1]:
        raise Exception("input matrix is not sqaure or not in rank 2")
    d = det(m)
    if d == 0:
        raise Exception("determinant of input matrix is zero, m is not invertable (singular)")
    adj = adjoint(m)
    return adj/d
    
```
