1. Fixed point iteration to solve the equation $x^4 - 2x - 10 = 0$

# Testing Idea of fix-point iteration

```python
def new_f1(x):
    return np.power(2*x + 10, 1./4.)
def new_f2(x):
    return -np.power(2*x + 10, 1./4.)
```

```python
r = 1
for i in range(10):
    print(r)
    r = new_f1(r)
```

```python
r = 1
for i in range(10):
    print(r)
    r = new_f2(r)
```

# Implementing

```python


def fix_point_iteration(g, x, sc=E_S, verbose=False, **kwargs):
    epsilon_a = float("inf")
    while sc < epsilon_a:
        new_x = g(x)
        epsilon_a = abs((new_x - x)/ new_x) * 100
        x = new_x
    return x
```

```python
# take 2 as first guess
root1 = fix_point_iteration(new_f1, 2, E_S)
root1
```

```python
# take -2 as first guess
root2 = fix_point_iteration(new_f2, -2, E_S)
root2
```

```python
f(root1) # -> 0
```

```python
f(root2) # -> 0
```

# Implementing

```python
def f(x):
    return x**4 - 2*x - 10
def g(x):
    return np.e ** x + x ** (1/2) - 2*x - 5

def newtons_iteration(f, x, sc=E_S, verbose=False, **kwargs):
    n = sympy.symbols("n")
    f_sym = f(n)
    df_sym = f_sym.diff()
    quotient = f_sym / df_sym
    quotient_lam = sympy.lambdify(n, quotient)
    epsilon_a = float("inf")
    while sc < epsilon_a:
        new_x = x - quotient_lam(x)
        epsilon_a = abs((new_x - x)/ new_x) * 100
        x = new_x
    return x
```

```python
root1 = newtons_iteration(g, 1.)
root1
```
