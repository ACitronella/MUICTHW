## 1. homework 5.1 maximum element of a list using the concept of dynamic programming (recursion)

```python
def find_max(l, cur=-float("inf")):
    if len(l) == 0:
        return cur
    if l[0] > cur:
        cur = l[0]
    return find_max(l[1:], cur)
```

## 2. homework 5.4 golden section search

```python
import math
phi = (1. + math.sqrt(5.))/ 2.
phi_1 = (phi - 1)
phi_2 = (2 - phi)
```

```python
def golden_section_search_min(f, xl, xu, sc=5E-4, verbose=False):
    """
    this func will return local minima in [xl, xu] of f
    """
    ea = float("inf")
    ret = float("inf")
    while sc < ea:
        d = phi_1 * (xu - xl)
        x1 = xl + d
        x2 = xu - d
        fx1 = f(x1)
        fx2 = f(x2)
        if verbose:
            print("x1 = %f\tf(x1) = %f\tx2 = %f\tf(x2) = %f" % (x1, fx1, x2, fx2))
        if fx1 < fx2:
            xl = x2
            ret = x1
        else:
            xu = x1
            ret = x2
        ea = phi_2 * abs((xu - xl) / ret) * 100
    return ret

def golden_section_search_max(f, xl, xu, sc=5E-4, verbose=False):
    """
    this func will return local minima in [xl, xu] of f
    """
    ea = float("inf")
    ret = float("inf")
    while sc < ea:
        d = phi_1 * (xu - xl)
        x1 = xl + d
        x2 = xu - d
        fx1 = f(x1)
        fx2 = f(x2)
        if verbose:
            print("x1 = %f\tf(x1) = %f\tx2 = %f\tf(x2) = %f" % (x1, fx1, x2, fx2))
        if fx1 > fx2:
            xl = x2
            ret = x1
        else:
            xu = x1
            ret = x2
        ea = phi_2 * abs((xu - xl) / ret) * 100
    return ret
```

```python
def f(x):
    return math.exp(x) - x**3
```

```python
golden_section_search_min(f, 2, 6, sc=0.005, verbose=True)
```

## 3. homework 5.5 parabolic interpolation

```python
def parabolic_interpolation_min(f, x1, x2, x3, sc=5E-4, verbose=False):
    ea = float("inf")
    xmin_old = float("inf")
    while sc < ea:
        x21 = x2 - x1
        fx23 = f(x2) - f(x3)
        x23 = x2 - x3
        fx21 = f(x2) - f(x1)
        x4 = x2 - 0.5 * (x21 * x21 * fx23 - x23 * x23 * fx21) / (x21 * fx23 - x23 * fx21)
        if x4 > x2:
            if f(x4) < f(x2):
                x1 = x2
                x2 = x4
                xmin = x4
            else:
                x3 = x4
                xmin = x2
        else:
            if f(x4) < f(x2):
                x3 = x2
                x2 = x4
                xmin = x4
            else:
                x1 = x4
                x4 = x2
                xmin = x2
                
        ea = abs((xmin - xmin_old) / xmin) * 100
        xmin_old = xmin
    return xmin

def parabolic_interpolation_max(f, x1, x2, x3, sc=5E-4, verbose=False):
    ea = float("inf")
    xmin_old = float("inf")
    while sc < ea:
        x21 = x2 - x1
        fx23 = f(x2) - f(x3)
        x23 = x2 - x3
        fx21 = f(x2) - f(x1)
        x4 = x2 - 0.5 * (x21 * x21 * fx23 - x23 * x23 * fx21) / (x21 * fx23 - x23 * fx21)
        if x4 > x2:
            if f(x4) > f(x2):
                x1 = x2
                x2 = x4
                xmin = x4
            else:
                x3 = x4
                xmin = x2
        else:
            if f(x4) > f(x2):
                x3 = x2
                x2 = x4
                xmin = x4
            else:
                x1 = x4
                xmin = x2
                
        ea = abs((xmin - xmin_old) / xmin) * 100
        xmin_old = xmin
    return xmin
        
```

## 1. homework 5.2 root-finding methods (for optimization)

```python
import sympy
import numpy as np
```

```python
# code from week 3
p = 4
E_S = 0.5 * 10 ** (2 - p)
def bisection_iter(f, x, xn, sc=E_S, verbose=False, **kwargs):
    epa = float("inf")
    current_root = 0.
    old_root=float("inf")
    n = 0
    if verbose:
        print("\tstart\t\tend\t\tcurrent root\tepsilon a")
    while sc < epa:
        current_root = (x + xn) / 2.
        epa = abs((current_root - old_root) / current_root) * 100
        if verbose:
            print("iter%d:\t%f\t%f\t%f\t%f" % (n, x, xn, current_root, epa))
        if f(x) * f(current_root) < 0.:
            xn = current_root
        else:
            x = current_root
        old_root = current_root
        n += 1
    return current_root

def roots_searching(algo, f, start:float, end:float, sc:float=E_S, interval_num:int=10, verbose=False) -> list:
    s = np.linspace(start, end, interval_num, endpoint=True)
    if verbose:
        print("searching in range [%f, %f] with %d sub interval and sc = %f" % (start, end, interval_num, sc))
    roots = []
    
    for x, xn in zip(s, s[1:]):
        if f(x) * f(xn) < 0:
            if verbose:
                print("ensure root in interval [%f, %f]" % (x, xn))
            root = algo(f, x, xn, sc, verbose=verbose, interval_num=interval_num)
            if verbose:
                print("approximate root as", root, "\n")
            roots.append(root)
    return roots
```

```python
def root_search_critical_point(f, xu, xl, sc=5E-4, verbose=False):
    n = sympy.symbols("n")
    f_sym = f(n)
    f_dot_sym = f_sym.diff()
    f_double_dot_sym = f_dot_sym.diff()
    f_dot_lam = sympy.lambdify(n, f_dot_sym)
    f_double_dot_lam = sympy.lambdify(n, f_double_dot_sym)
    if verbose:
        print("first derivative:", f_dot_sym)
        print("second derivative:", f_double_dot_sym)
    
    roots = roots_searching(bisection_iter, f_dot_lam, xu, xl, sc, verbose=verbose)
    for r in roots:
        if f_double_dot_lam(r) > 0.:
            print("min: ", end="")
        elif f_double_dot_lam(r) < 0.:
            print("max: ", end="")
        else:
            print("critical point: ", end="")
        print(r)
    return roots
```

```python
# according to wolframalpha, approximate solution is 3.6330
# according to wolframalpha, approximate solution is 0.36701
root_search_critical_point(f1, -2, 6)
```

## 2. homework 5.3 equal interval search

```python
def equal_interal_search_min(f, xl, xu, sc=5E-4, verbose=False):
    """
    this func will return local minima in [xl, xu] of f
    """
    ea = float("inf")
    xm_old = float("inf")
    while sc < ea:
        xm = (xl + xu) * 0.5
        d = (xu - xl) * 0.01
        x1 = xm - d
        x2 = xm + d
        fx1 = f(x1)
        fx2 = f(x2)
        if verbose:
            print("x1 = %f\tf(x1) = %f\tx2 = %f\tf(x2) = %f" % (x1, fx1, x2, fx2))
        if fx1 < fx2:
            xu = x2
        else:
            xl = x1
        ea = abs((xm - xm_old) / xm) * 100
        xm_old = xm
    return xm

def equal_interal_search_max(f, xl, xu, sc=5E-4, verbose=False):
    """
    this func will return local minima in [xl, xu] of f
    """
    ea = float("inf")
    xm_old = float("inf")
    while sc < ea:
        xm = (xl + xu) * 0.5
        d = (xu - xl) * 0.01
        x1 = xm - d
        x2 = xm + d
        fx1 = f(x1)
        fx2 = f(x2)
        if verbose:
            print("x1 = %f\tf(x1) = %f\tx2 = %f\tf(x2) = %f" % (x1, fx1, x2, fx2))
        if fx1 < fx2:
            xl = x1
        else:
            xu = x2
        ea = abs((xm - xm_old) / xm) * 100
        xm_old = xm
    return xm
```
